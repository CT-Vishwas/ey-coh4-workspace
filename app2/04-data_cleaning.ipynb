Data Cleaning
----------------
1. **Handling Missing Values**: Use `pd.isnull()` to identify missing values and `pd.fillna()` to fill them with a specified value or method (e.g., forward fill, backward fill).
2. **Removing Duplicates**: Use `pd.drop_duplicates()` to remove duplicate rows from the DataFrame.
3. **Data Type Conversion**: Use `pd.to_numeric()`, `pd.to_datetime()`, or `pd.astype()` to convert data types of columns as needed.

```python
# Dictionary to create Series
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
}
```

```python
# Dictionary to create DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
# generate a dictionary with 100 entries to demonstrate data cleaning
data = {
    'Name': ['Person ' + str(i) for i in range(1, 101)],
    'Age': [np.random.choice([25, 30, 35, 40, 45, None]) for _ in range(100)],
    'City': [np.random.choice(['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', None]) for _ in range(100)]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

1. dropna() - Remove rows with missing values
2. fillna() - Fill missing values with a specified value or method
3. duplications - Remove duplicate rows

```python
df1 = df.copy()
df1.dropna(inplace=True)
df1.info()
```

```python
df2 = df.copy()
df2.fillna({'Age': df2['Age'].mean(), 'City': 'Unknown'}, inplace=True)
df2.info()
```

```python
# Example data with wrong data types
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': ['25', '30', '35', '40', '45'],  # Age as strings
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df.info()
```

```python
df1 = df.copy()
df1['Age'] = df1['Age'].astype(int)
df1.info()
```

```python
# data with wrong date format
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Birthdate': ['01-01-1990', '02/02/1985', '03-03-1980', '04-04-1975', '05-05-1970']  # Wrong format
}
df = pd.DataFrame(data)
print(df)
```

```python
# Fix the date format
df['Birthdate'] = pd.to_datetime(df['Birthdate'], format='%d-%m-%Y')
print(df)
```

```python
# Extract date from Bob user
bob_birthdate = df.loc[df['Name'] == 'Bob', 'Birthdate'].iloc[0]
print(f"Bob's Birthdate: {bob_birthdate}")
# Fix the date format of Bob's birthdate from '02/02/1985' to '02-02-1985'
bob_birthdate_fixed = pd.to_datetime(bob_birthdate, format='%d-%m-%Y')
print(f"Bob's Birthdate after fixing format: {bob_birthdate_fixed}")
# Now we can update the DataFrame with the fixed birthdate for Bob
df.loc[df['Name'] == 'Bob', 'Birthdate'] = bob_birthdate_fixed
```
